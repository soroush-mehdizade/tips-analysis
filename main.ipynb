# Relationship Between Total Bill and Tip Amount
The scatter plot also confirms what we said: as the total bill rises, the tips increase with it.

```python
px.scatter(data_frame=df ,x = 'total_bill' ,y = 'tip' ,size='tip',title='Relationship Between Total Bill and Tip Amount',
           height=600,width=900)
```

# Relationship Between Size and Tip Amount
Plotting the number of guests against the tip amount also supports our point — we can see that as the number of people increases, the total tip amount tends to increase as well


```python
df_sum = df.groupby('size', as_index=False)['tip'].mean()


fig = px.bar(df_sum, x='size', y='tip', title='Total Tips by Table Size',height=600,width=900)
fig.show()
```

# The number of tips given by men and women 
Let’s go a bit further and compare the tipping behavior of men and women.
When we plot the total tips by gender using a bar chart, we can see that men tend to leave more tips than women.
This may be because men often act as the head of the household, or because they are more likely to pay the tip during a two-person date to impress their companion.

```python
df_sum_2 = df.groupby('sex' ,as_index = False)['tip'].count()

px.bar(data_frame=df_sum_2 ,x = 'sex' ,y = 'tip',color='sex',title='The number of tips given by men and women',height=600,width=900)
```

# Impact of group size and smoking status on tip count
As we can see, men tend to tip more generously when they are in a group than when they are alone. In particular, during two-person dates, men often leave larger tips, possibly to impress their companion.
Women, on the other hand, also tend to tip when dining alone but are generally less motivated by the desire to impress others.

Another possible reason why women give smaller tips on average is financial: in many societies, men are more likely to belong to wealthier income groups, while women tend to have lower average incomes.

In addition, the plot shows that smokers usually tip less than non-smokers. One explanation could be that smokers are more likely to belong to middle or lower income brackets, while wealthier individuals tend to smoke less, live healthier lifestyles, and perhaps show better social behavior — which may include tipping more generously.

```python
fig = px.histogram(
    df,
    x='sex',
    facet_col='size',
    category_orders={'size':[1,2,3,4,5,6]},
    text_auto=True  ,
    color='sex',
    height=600,width=900,
    facet_row='smoker'
)

fig.show()
```

# Impact of weekday and time of day on tipping
Now let’s examine the impact of the day of the week and the time of day on tipping behavior.
By plotting the tips across different days and times, we can see that the highest tips are usually given on Saturday evening and Sunday evening.

One possible explanation is that Thursday is still a workday, so people often visit restaurants for business lunches rather than dinner.
In contrast, Saturday and Sunday are weekend days, when people are off work and more likely to go out for evening meals, which naturally leads to higher tips.

```python
fig = px.histogram(
    df,
    x='sex',
    facet_col='day',
    category_orders={'day':['Thur' ,'Fri' ,'Sat' ,'Sun'] ,'time':['Lunch' ,'Dinner']},
    text_auto=True  ,
    color='sex',
    height=600,width=900,
    facet_row='time'
)

fig.show()
```
